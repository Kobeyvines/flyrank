## 2. The question: decision, action, cost of a wrong call

**Search Question:** What safe content features and search signals show measurable, statistical associations with organic visibility and clicks?

**Unit of Analysis:** A single pseudonymized URL/Page represented by a content identifier over a historical aggregation window.

**Decision and Action:** Content leads will use this work to decide which content creation standards and optimization guidelines are worth promoting. The concrete output is a "Signal Audit & Content Strategy Guidelines" report that can help teams prioritize which signals are worth testing and which ones should be treated as noise.

**Cost of a Wrong Call:** If the team follows weak or spurious associations, it can waste editorial budget on content changes that do not improve visibility. The opposite error is also costly: treating an observed correlation as direct causality can lead teams to misread the evidence and ignore signals that may be useful for decision support.

- The dataset contains 30,000 content pages and 32 pseudonymized clients, which gives enough scale for a meaningful observational audit.
- The observed correlation between search volume and 90-day impressions is 0.0012, which is extremely small and suggests that raw demand alone is not a strong standalone signal.
- The average page word count is 3,107.8 words, which is a useful baseline for questioning whether length is an automatic quality signal.


````python
import os
import pandas as pd
import numpy as np

candidate_paths = [
    "data/raw/content_refresh_anonymized.csv",
    "../../data/raw/content_refresh_anonymized.csv",
]

file_path = next((p for p in candidate_paths if os.path.exists(p)), None)
if file_path is None:
    raise FileNotFoundError("Could not find the dataset in the expected locations.")

df = pd.read_csv(file_path)

row_count = len(df)
client_count = df['client_id'].nunique() if 'client_id' in df.columns else np.nan
volume_corr = df['search_volume'].corr(df['impressions_90d']) if 'search_volume' in df.columns and 'impressions_90d' in df.columns else np.nan
avg_word_count = df['word_count'].mean() if 'word_count' in df.columns else np.nan

print(f"Dataset loaded from: {file_path}")
print(f"Total Content Pages Analyzed: {row_count:,}")
print(f"Pseudonymized Clients: {client_count}")
print(f"Correlation (Search Volume vs. 90-day Impressions): {volume_corr:.4f}")
print(f"Average Word Count across pages: {avg_word_count:.1f} words")
````
