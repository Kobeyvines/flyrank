## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

*   Supporting Data Highlights (Computed Live from content\_refresh\_anonymized.csv):
    

*   **Dataset Scope:** **30,000** anonymized page records evaluated across **32** pseudonymized client domains.
    
*   **Search Volume Reality Check:** Search volume exhibits a negligible linear correlation of **0.0012** with actual 90-day impressions. High demand does not equal organic visibility without execution and strong placement.
    
*   **Word Count Baseline:** The average page length across the entire corpus is **3,107.8 words**. Word count alone shows negligible differentiation between top and bottom performers, demonstrating that "write longer content" is an oversimplified heuristic.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os
import pandas as pd
import numpy as np

# Path fallback check
file_path = "../../data/raw/content_refresh_anonymized.csv"
if not os.path.exists(file_path):
    file_path = "../data/raw/content_refresh_anonymized.csv"

# Load the starter dataset
df = pd.read_csv(file_path)

# 1. Total volume & scope
total_pages = len(df)
total_clients = df['client_id'].nunique() if 'client_id' in df.columns else "N/A"

# 2. Key signal check: Search Volume vs. 90-day Impressions
volume_corr = df['search_volume'].corr(df['impressions_90d']) if 'search_volume' in df.columns and 'impressions_90d' in df.columns else np.nan

# 3. Word Count distribution check
avg_word_count = df['word_count'].mean() if 'word_count' in df.columns else np.nan

print(f"--- LANE 1 STARTER DATA AUDIT ---")
print(f"Dataset successfully loaded from: {file_path}")
print(f"Total Content Pages Analyzed: {total_pages:,}")
print(f"Pseudonymized Clients: {total_clients}")
print(f"Correlation (Search Volume vs. 90d Impressions): {volume_corr:.4f}")
print(f"Average Word Count across pages: {avg_word_count:.1f} words")
```
